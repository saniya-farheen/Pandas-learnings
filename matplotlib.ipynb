```python
import matplotlib.pyplot as plt
```

```python
x = [1,2,3,4,5,6]
y = [20,30,50,10,70,40]
plt.plot(x,y)
```

```python
# adding the title
plt.plot(x,y)
plt.title("marks according to id")
plt.xlabel('id')
plt.ylabel('marks')
```

```python
# changing colors and style 
plt.plot(x,y, color='pink', linestyle='--' , marker = '*') 
plt.show()
```

```python
#bar chart 
students = ['a','b','c','d']
marks = [50,70,90,40]

plt.bar(students, marks, color='green')
plt.show()
```

```python
#Histogram 

ages = [20,18,13,12,24,50,55,44,32,70]

plt.hist(ages)
plt.show()
```

```python
#scatter plot

height = [5,6,7,4,3,8]
weight=[60,90,120,40,20,140]
plt.scatter(height, weight)
plt.show()
```

```python
#pie chart 
sizes = [40,30,20,10]
id = ['A','B','C','D']

plt.pie(sizes,labels = id)
plt.show()
```

```python
# complete matplotlib

months = ['jan', 'feb', 'mar', 'apr', 'may', 'jun', 'jul', 'aug', 'sep', 'oct', 'nov', 'dec']

sales= [100,200,150,400,300,550,450,300,260,540,430,120]
plt.figure(figsize=(15,20))

plt.plot(months, sales ,color='green', linestyle= '--', marker= '*')
plt.xticks(rotation=45) # to rotate the x axis label
plt.yticks(rotation=45) # to rotate the y axis label
plt.title('year sale')
plt.xlabel(' sale months')
plt.ylabel('sale cost')


plt.grid(color = 'grey', linestyle = '--', linewidth = 0.5)

```
